```python
# Remove duplicate medicine names
df = df.drop_duplicates(subset="Medicine Name", keep="first").reset_index(drop=True)

print("New Shape:", df.shape)
print("Duplicate Medicine Names:", df["Medicine Name"].duplicated().sum())
```

```python
# Create text column for Transformer
df["medicine_text"] = df["Medicine Name"].astype(str).str.strip()

print(df[["Medicine Name", "medicine_text"]].head())
```

```python
print("Total medicines:", len(df))
print("Total text values:", df["medicine_text"].notna().sum())
```

```python
from sentence_transformers import SentenceTransformer

print("Loading Transformer model...")

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print("Transformer loaded successfully!")
```

```python
# Get medicine names
medicine_names = df["medicine_text"].tolist()

print("Total medicines:", len(medicine_names))

# Generate embeddings
embeddings = model.encode(
    medicine_names,
    show_progress_bar=True,
    batch_size=32
)

print("Embedding shape:", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def search_medicine(query, top_k=5):

    # Convert user query into embedding
    query_embedding = model.encode([query])

    # Compare query with combined medicine information
    similarities = cosine_similarity(
        query_embedding,
        combined_embeddings
    )[0]

    # Get top matches
    top_indices = similarities.argsort()[-top_k:][::-1]

    results = df.iloc[top_indices].copy()

    results["Similarity"] = similarities[top_indices]

    return results
```

```python
results = search_medicine("Augmentin 625", top_k=5)

display(
    results[
        [
            "Medicine Name",
            "Composition",
            "Uses",
            "Side_effects",
            "Similarity"
        ]
    ]
)
```

```python
def get_medicine_info(query):

    results = search_medicine(query, top_k=1)

    medicine = results.iloc[0]

    print("\n" + "=" * 60)
    print("MEDICINE INFORMATION")
    print("=" * 60)

    print("Medicine Name :", medicine["Medicine Name"])
    print("Composition   :", medicine["Composition"])
    print("Uses          :", medicine["Uses"])
    print("Side Effects  :", medicine["Side_effects"])
    print("Manufacturer  :", medicine["Manufacturer"])

    print("\nReview:")
    print("Excellent :", medicine["Excellent Review %"], "%")
    print("Average   :", medicine["Average Review %"], "%")
    print("Poor      :", medicine["Poor Review %"], "%")

    print("\nSimilarity Score :", round(medicine["Similarity"], 4))
```
